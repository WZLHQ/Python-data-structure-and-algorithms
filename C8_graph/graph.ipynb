# 图（graph）
每一个节点有多个前驱与多个后继，节点之间是多对多的关系，可以表示具有各种复杂联系的数据集合。相关术语如下：
* 顶点（Vertex，也称节点Node）：是图的基本组成部分，顶点具有名称标识Key，也可以携带数据项payload。
* 边（Edge，也称弧Arc）：作为2个顶点之间关系的表示，边连接两个顶点；边可以是无向或者有向的，相应的图称为“无向图”和“有向图”。
* 权重Wieght：为了表达从一个顶点到另一个顶点的“代价”，可以给边赋权；例如公交网络中两个站点之间的“距离”、“通行时间”和“票价”都可以作为权重。具有权重的图称为赋权图。
* 一个图G可以定义为G=(V,E)：其中V是顶点的集合，E是边的集合，E中的每条边e=(v,w)。一个赋权图的例子如下：

![赋权图例子](../images/赋权图例子.png)

* 路径Path：无权路径的长度为边的数量，带权路径的长度为所有边权重的和。
* 圈Cycle：圈是首尾顶点相同的路径。如果有向图中不存在任何圈，则称作“有向无圈图directed acyclic graph: DAG”。树可以看成一种DAG。

ADT Graph的实现方法有两种主要形式：
* 邻接矩阵adjacency matrix
* 邻接表adjacency list

这两种方法各有优劣，需要在不同应用中加以选择，一般来说邻接矩阵适合于稠密图，邻接表适合于稀疏图。



```python
# 利用邻接矩阵实现简单图
class GraphAX:
    '''
    利用邻接矩阵实现简单图时，图的顶点数据量是确定的，一般不讨论增加顶点的功能。
    用0初始化邻接矩阵。
    '''
    def __init__(self,num_vertices, directed=False):
        
        self.num_vertices=num_vertices
        self.directed=directed

        # 初始化邻接矩阵
        self.init_AX=[ [0]*num_vertices for _ in range(num_vertices)]
    

    # 以下为图基本操作
    def add_edge(self,v1,v2,weight=1):
        if v1 in range(self.num_vertices) and v2 in range(self.num_vertices):
            self.init_AX[v1][v2]=weight
            if not self.directed:
                self.init_AX[v2][v1]=weight
        else:
            print(f"错误：顶点索引超出范围({v1},{v2})")

    def remove_edge(self,v1,v2):
        if v1 in range(self.num_vertices) and v2 in range(self.num_vertices):
            self.init_AX[v1][v2]=0
            if not self.directed:
                self.init_AX[v2][v1]=0
        else:
            print(f"错误：顶点索引超出范围({v1},{v2})")

    def has_edge(self,v1,v2):
        if v1 in range(self.num_vertices) and v2 in range(self.num_vertices):
            return self.init_AX[v1][v2]!=0
        return False

    def get_weight(self,v1,v2):
        if v1 in range(self.num_vertices) and v2 in range(self.num_vertices):
            return self.init_AX[v1][v2]
        return None
    

    # 以下为图的查询功能
    def get_neighbors(self,v):
        '''获取某个顶点的所有邻接顶点及其权重'''
        if v in range(self.num_vertices):
            neighbors=[]
            for id,i in enumerate(self.init_AX[v]):
                if i!=0:
                    neighbors.append((id,i))
            return neighbors
        return []

    def get_degree(self,v):
        '''获取顶点的度'''
        if self.directed:
            out_degree=sum(1 for i in self.init_AX[v][:] if i!=0)
            in_degree=sum(1 for row in self.init_AX if row[v]!=0)
            return out_degree+in_degree
        else:
            return sum(1 for i in self.init_AX[v][:] if i!=0)

    def display(self):
        """显示邻接矩阵"""
        print("邻接矩阵：")
        print("   ", end="")
        for i in range(self.num_vertices):
            print(f"{i:3}", end=" ")
        print()
        
        for i in range(self.num_vertices):
            print(f"{i:2}:", end=" ")
            for j in range(self.num_vertices):
                print(f"{self.init_AX[i][j]:3}", end=" ")
            print()

    # 以下为图的遍历功能
    def dfs(self,):
        '''
        深度优先搜索
        '''
        pass

    def bfs(self,):
        '''
        广度有限搜索
        '''
        pass

if __name__=="__main__":
    # 创建一个有5个顶点的无向图
    g = GraphAX(5, directed=False)
    
    # 添加边
    g.add_edge(0, 1, 2)  # 顶点0和1之间的权重为2
    g.add_edge(0, 2, 4)  # 顶点0和2之间的权重为4
    g.add_edge(1, 2, 1)  # 顶点1和2之间的权重为1
    g.add_edge(1, 3, 3)  # 顶点1和3之间的权重为3
    g.add_edge(2, 4, 5)  # 顶点2和4之间的权重为5
    
    # 显示图
    g.display()
    
    print(f"\n顶点1的邻接点: {g.get_neighbors(1)}")
    print(f"顶点2的度: {g.get_degree(2)}")
    print(f"顶点0和1之间是否存在边: {g.has_edge(0, 1)}")
    print(f"顶点0和1之间的权重: {g.get_weight(0, 1)}")
    
    # 深度优先搜索
    # print(f"DFS从顶点0开始: {g.dfs(0)}")
    
    # 广度优先搜索
    # print(f"BFS从顶点0开始: {g.bfs(0)}")
    
    # 删除一条边
    g.remove_edge(1, 2)
    print(f"\n删除边(1,2)后:")
    g.display()

```
