```python
def removeElement(nums, val):
    """
    :type nums: List[int]
    :type val: int
    :rtype: int
    """
    count=0
    for ind, _ in enumerate(nums):
        while ind<=len(nums) and nums[ind]==val:
            nums.pop(ind)
            count+=1
    return count

# nums=[0,1,2,2,2,2,2,3,2,2,5]
nums=[3,2,2,3]
print("count",removeElement(nums,2))
print("nums",nums)
```
