# 排序算法的稳定性
稳定排序算法会让原本有相等键值的记录维持相对次序。也就是说一个排序算法是稳定的，当有两个相等键值的记录R和S，且在原本的列表中R出现在S之前，在排序过的列表中R也将会是在S之前。

# 冒泡排序及其实现

```python
def bubble_sort(alist):
    n=len(alist)
    for i in range(n):
        temp=0
        for j in range(n-i-1):
            A=alist[j]
            B=alist[j+1]
            if A>B:
                temp=A
                A=B
                B=temp
    print(alist)
bubble_sort([3,2])
```
