# 排序算法的稳定性
稳定排序算法会让原本有相等键值的记录维持相对次序。也就是说一个排序算法是稳定的，当有两个相等键值的记录R和S，且在原本的列表中R出现在S之前，在排序过的列表中R也将会是在S之前。

# 冒泡排序及实现

```python
# 无序表顺序查找
def sequentialSearch(alist,item):

    found=False

    for i in alist:
        if i==item:
            found=True
            break
    
    return found
        
print(sequentialSearch([1,2,3,4],1))

# 有序表如何进行查找
```

## 二分查找（针对有序表）


```python
# 二分查找

def binarySearch(alist,item):
    si=0
    ei=len(alist)-1
    found=False
    
    while si<=ei and not found:
        mi=(si+ei)//2
        if alist[mi]==item:
            found=True
        else:
            if alist[mi] > item:
                ei=mi-1
            else:
                si=mi+1

    return found

print(binarySearch([1,2,3,4,5,6],9))
```
