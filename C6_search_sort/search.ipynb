# 搜索
搜索是在一个项目集合中找到一个特定项目的算法过程。搜索通常的答案是真或假，因为该项目是否存在。搜索的几张常见方法：顺序查找、二分查找、二叉树查找、哈希查找

# 1 二分法查找
二分查找又称折半查找，优点是比较次数少，查找速度块；其缺点是要求待查表为有序表，且插入删除苦难。因此，二分查找方法适用于不经常变动而查找频繁的有序列表。首先，假设表中元素是按升排列，将表中间位置记录的关键字与查找关键字比较，如果两者相等，则查找成功；否则利用中间位置记录将表分成前后两个子表，如果中间位置记录的关键字大于查找关键字，则进一步查找前一子表，否则进一步查找后一字表。重复以上过程，知道找到满足条件的记录，使查找成功，或者直到子表不存在位置，此时查找不成功。

* 最优时间复杂度：O(1)
* 最坏时间复杂度：O(logn)

```python
# 非递归实现
def binary_search(alist,target):

    start,end=0,len(alist)-1

    while start<=end:
        mid=(end+start)//2
        if alist[mid]>target:
            end=mid-1
        elif alist[mid]<target:
            start=mid+1
        else:
            return True
        
    return False

print(binary_search([1,2,5,7,9],5))
```

```python
# 递归实现
def binary_search(alist,target):

    # 框架基本正确
    # 重点理解下如何将正确结果返回，或者理解下这个返回值的逻辑


    start,end=0,len(alist)-1
    mid=(start+end)//2
    FIND=False

    # 基本结束条件
    if len(alist)==[]:
        return
    
    # 基本问题
    if alist[mid]>target:
        binary_search(alist[start:mid],target)
    elif alist[mid]<target:
        binary_search(alist[mid+1:end+1],target)
    else:
        FIND=True
```
