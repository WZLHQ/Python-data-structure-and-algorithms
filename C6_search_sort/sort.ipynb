# 排序算法的稳定性
稳定排序算法会让原本有相等键值的记录维持相对次序。也就是说一个排序算法是稳定的，当有两个相等键值的记录R和S，且在原本的列表中R出现在S之前，在排序过的列表中R也将会是在S之前。

# 1 冒泡排序及其实现
冒泡排序（bubble sort）是一种简单的排序算法。它重复地遍历要排序地数列，一次比较两个元素，如果他们的顺序错误就把他们交换过来。遍历数列地工作是重复地进行，直到不需要再交换，即该数列已经排序完成。这个算法的名字由来是因为越小/大的元素会经由交换慢慢“浮”到数列的顶端。

冒泡排序的核心思想是：
* 通过循环地进行 相邻元素之间的换位操作 将数值最大的元素移动到尾端

冒泡排序的时间复杂度：
* 最坏时间复杂度为O(n**2)
* 最优时间复杂度为O(n)
* 满足稳定性


```python
# 将列表按照由小到大的顺序排列

def bubble_sort_1(alist):
    # 非递归解法
    n=len(alist)
    for i in range(n-1):
        # count的加入会降低冒泡排序算法的最优时间复杂度(可以提前结束外层循环)
        # 但是最坏时间复杂度不会改变
        count=0
        for j in range(n-i-1):
            if alist[j] > alist[j+1]:
                alist[j], alist[j+1] = alist[j+1], alist[j]
                count+=1
        if count==0:
            break
    return alist

def bubble_sort_2(alist,start,end):
    # 递归解法
    # 长序列的冒泡排序不适合用递归算法，此处只是为了加深对递归的理解

    # 基本结束条件
    if start==end:
        return

    # 基本问题
    for i in range(start,end):
        if alist[i]>alist[i+1]:
            alist[i],alist[i+1]=alist[i+1],alist[i]

    # 调用自身以缩小问题规模
    bubble_sort_2(alist,start,end=end-1)

if __name__ == "__main__":

    # 测试非递归解法
    li=[3,1,3,3,1,1,2,8,7,-1,-5,-2,-5]
    print(bubble_sort_1(li))

    # 测试递归解法
    li=[3,1,3,3,1,1,2,8,7,-1,-5,-2,-5]
    bubble_sort_2(li,0,len(li)-1)
    print(li)
    
```

# 2 选择排序及其实现
选择排序（selection sort）是一种简单直观的排序算法。其原理为：首先在未排序序列中找到最小/大元素，存放到排序序列的起始位置；然后，再从余下序列中继续寻找最小/大元素，然后放到已排序序列末尾。以此类推，直到所有元素均排列完毕。

选择排序的核心思想是：
* 首次将第一个元素与剩下的序列划分为两部分，同时将第一个元素替换为剩下序列中的最大/小元素，则第一个元素为有序部分，剩下为无序部分；
* 然后对无序部分重复进行上述步骤
* 主要解决的问题是：如何从无序序列中找到目标元素

选择排序的时间复杂度：
* 最坏与最坏时间复杂度均为O(n**2)
* 不满足稳定性（考虑升序每次选择最大的情况）

```python
# 将列表按照由小到大的顺序排列

def selection_sort_1(alist):
    # 内层循环每次都进行交换-->代码更优雅些，但是需要内存开销支持频繁交换
    n=len(alist)
    i=0
    while i<n-1:
        for j in range(i+1,n):
            if alist[i]>alist[j]:
                alist[i], alist[j] = alist[j], alist[i]
        i+=1
    return alist

def selection_sort_2(alist):
    # 内层循环只进行一次交换-->代码稍微复杂些（蕴含了找最大值下标的算法），但是避免频繁交换
    n=len(alist)
    i=0
    while i<n-1:
        min_index=i
        for j in range(i+1,n):
            if alist[min_index]>alist[j]:
                min_index=j
        alist[i], alist[min_index] = alist[min_index], alist[i]
        i+=1
    return alist

if __name__ == "__main__":
    # 测试第一种优雅实现
    li=[4,3,15,6,8,-1000,12,-5,0]
    print(selection_sort_1(li))

    # 测试第二种高效实现
    li=[4,3,15,6,8,-1000,12,-5,0]
    print(selection_sort_1(li))
```

# 3 插入排序及其实现
插入排序（insertion sort）是一种简单直观的排序算法。它的工作原理是通过构建有序序列，对于未排序数据，在已排序序列中从后向前扫描，找到相应位置并插入。

插入排序的核心思想是：
* 如何将无序序列中的元素逐个插入到有序序列中，并且如何优化插入算法的时间复杂度

插入排序的时间复杂度为：
* 最坏时间复杂度为O(n^2)
* 最优时间复杂度为O(n)
* 满足稳定性

```python
# 将列表按照由小到大的顺序排列
def inerst_sort_1(alist):
    # 内层循环使用while
    n=len(alist)
    for i in range(n-1):
        while i>=0:
            if alist[i]>alist[i+1]:
                alist[i],alist[i+1]=alist[i+1],alist[i]
                i-=1
            else:
                break # 优化插入算法的时间复杂度
    return alist

def inerst_sort_2(alist):
    # 内层循环使用for
    n=len(alist)
    for i in range(n-1):
        for j in range(i,-1,-1):
            if alist[j]>alist[j+1]:
                alist[j],alist[j+1]=alist[j+1],alist[j]
            else:
                break # 优化插入算法的时间复杂度
    return alist

if __name__=="__main__":
    li=[3,4,1,6,3,7,2]
    print(inerst_sort_1(li))

    li=[3,4,1,6,3,7,2]
    print(inerst_sort_2(li))
```

# 5 快速排序及其实现
快速排序（quick sort）

快速排序算法主要解决的问题是：

快速排序的时间复杂度为：
* 最坏时间复杂度为：O(n^2)
* 最优时间复杂度为：O(nlogn)
* 是否满足稳定性：不稳定

```python
# 快速排序算法
def quick_sort(alist,start,end):

    # 基本结束条件
    # TODO 这里的基本技术条件如果写为start==end，则会报错，思考为什么？
    if start>=end:
        return
    
    # 基本问题
    low,high=start,end
    mid_value=alist[low]
    while low<high:

        while alist[high]>=mid_value and low<high:
            high-=1
        alist[low]=alist[high]

        while alist[low]<mid_value and low<high:
            low+=1
        alist[high]=alist[low]
    
    alist[low]=mid_value

    # 调用自身
    quick_sort(alist,start,low-1)
    quick_sort(alist,low+1,end)

if __name__ == "__main__":
    li=[-1000,-22324234234,-223,1,3,2,8,3,2,7,9,13,10,-1,-2,3,5,7,9,12,3,465,7,8,1,1,1,1,1,1]
    quick_sort(li,0,len(li)-1)
    print(li)
```

# 6 归并排序及其实现
归并排序（merge sort）

归并排序算法主要解决的问题是：
* 如何

快速排序的时间复杂度为：
* 最坏时间复杂度为：O(nlogn)
* 最优时间复杂度为：O(nlogn)
* 是否满足稳定性：稳定

```python

def merge_sort(alist):

    n=len(alist)

    # 基本结束条件
    if n==1:
        return alist
    
    # 基本问题
    mid=n//2
    
    # 调用自身减小问题规模
    # 需要将left_list与left_right认为都是排序完成的结果
    left_list=merge_sort(alist[0,mid])
    right_list=merge_sort(alist[mid,n])

    # 合并
    result_list=[]
    index_left, index_right=0,0
    
    while index_left<=mid or index_right<=n: # ?
        if left_list[index_left]<right_list[index_right]:
            result_list.append(left_list[index_left])
            index_left+=1
        else:
            result_list.append(right_list[index_right])
            index_right+=1
    
    return result_list

```
