# 排序算法的稳定性
稳定排序算法会让原本有相等键值的记录维持相对次序。也就是说一个排序算法是稳定的，当有两个相等键值的记录R和S，且在原本的列表中R出现在S之前，在排序过的列表中R也将会是在S之前。

# 冒泡排序及其实现
* 最坏时间复杂度为O(n**2)
* 满足稳定性

```python
# 将列表按照由小到大的顺序排列
def bubble_sort(alist):
    '''
    最坏时间复杂度为O(n**2)
    count的加入会降低冒泡排序算法的最优时间复杂度(可以提前结束外层循环)
    但是最坏时间复杂度不会改变
    '''
    n=len(alist)
    for i in range(n-1):
        count=0
        for j in range(n-i-1):
            if alist[j] > alist[j+1]:
                alist[j], alist[j+1] = alist[j+1], alist[j]
                count+=1
        if count==0:
            break
    print(alist)

# 将列表按照由大到小的顺序排列?


bubble_sort([1,2,3,4])
```

# 选择排序及其实现
* 最坏时间复杂度为O(n**2)
* 不满足稳定性（考虑升序每次选择最大的情况）

```python
# 将列表按照由小到大的顺序排列
def select_sort(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(1+i,n):
            if alist[j]<alist[i]:
                alist[i], alist[j]=alist[j], alist[i]
    print(alist)

# 将列表按照由大到小的顺序排列?

select_sort([2,5,3,1,1])
```

```python
# 将列表按照由小到大的顺序排列
def insert_sort(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(i+1,0,-1):
            if alist[j]<alist[j-1]:
                alist[j],alist[j-1]=alist[j-1],alist[j]
    print(alist)

# 将列表按照由大到小的顺序排列?

insert_sort([5,3,1,2,-1,-9])
```
