# 排序算法的稳定性
稳定排序算法会让原本有相等键值的记录维持相对次序。也就是说一个排序算法是稳定的，当有两个相等键值的记录R和S，且在原本的列表中R出现在S之前，在排序过的列表中R也将会是在S之前。

# 1 冒泡排序及其实现
冒泡排序（bubble sort）是一种简单的排序算法。它重复地遍历要排序地数列，一次比较两个元素，如果他们的顺序错误就把他们交换过来。遍历数列地工作是重复地进行，直到不需要再交换，即该数列已经排序完成。这个算法的名字由来是因为越小的元素会经由交换慢慢“浮”到数列的顶端。

冒泡排序算法的核心思想是：
* 通过循环地进行 相邻元素之间的换位操作 将数值最大的元素移动到尾端

冒泡排序的时间复杂度：
* 最坏时间复杂度为O(n**2)
* 满足稳定性


```python
# 将列表按照由小到大的顺序排列
def bubble_sort_1(alist):
    '''
    最坏时间复杂度为O(n**2)
    count的加入会降低冒泡排序算法的最优时间复杂度(可以提前结束外层循环)
    但是最坏时间复杂度不会改变
    '''
    n=len(alist)
    for i in range(n-1):
        count=0
        for j in range(n-i-1):
            if alist[j] > alist[j+1]:
                alist[j], alist[j+1] = alist[j+1], alist[j]
                count+=1
        if count==0:
            break
    print(alist)

# 将列表按照由大到小的顺序排列
# TODO 如何提升算法的效率？
def bubble_sort_2(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(n-1-i):
            if alist[j]<alist[j+1]:
                alist[j],alist[j+1]=alist[j+1],alist[j]
    print(alist)

alist=[4,3,2,10]
bubble_sort_1(alist)
bubble_sort_2(alist)
```

# 2 选择排序及其实现
选择排序（selection sort）是一种简单直观地排序算法。其原理为：首先再未排序序列中找到最小/大元素，存放到排序序列的起始位置；然后，再从余下序列中继续寻找最小/大元素，然后放到已排序序列末尾。以此类推，直到所有元素均排列完毕。

选择排序算法的核心思想是：
* 首次将第一个元素与剩下的序列划分为两部分，同时将第一个元素替换为剩下序列中的最大/小元素，则第一个元素为有序部分，剩下为无序部分；
* 然后对无序部分重复进行上述步骤
* 主要解决的问题是：如何从无序序列中找到目标元素

选择排序的时间复杂度：
* 最坏时间复杂度为O(n**2)
* 不满足稳定性（考虑升序每次选择最大的情况）

```python
# 将列表按照由小到大的顺序排列
def select_sort_1(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(1+i,n):
            if alist[j]<alist[i]:
                alist[i], alist[j]=alist[j], alist[i]
    print(alist)

# 将列表按照由大到小的顺序排列
def select_sort_2(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(i,n-1):
            if alist[i]<alist[j+1]:
                alist[i],alist[j+1]=alist[j+1],alist[i]
    print(alist)

alist=[4,3,2,10,101,-100]
select_sort_1(alist)
select_sort_2(alist)
```

# 3 插入排序及其实现
插入排序（insertion sort）是一种简单直观的排序算法。它的工作原理是通过构建有序序列，对于未排序数据，在已排序序列中从后向前扫描，找到相应位置并插入。

插入排序算法的核心思想是：
* 1
* 2
* 主要解决的问题是：如何将无序序列中的元素逐个插入到有序序列中

插入排序的时间复杂度为：
* 最坏时间复杂度为O(n^2)
* 是否满足稳定性？

```python
# 将列表按照由小到大的顺序排列
def insert_sort_1(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(i+1,0,-1):
            if alist[j]<alist[j-1]:
                alist[j],alist[j-1]=alist[j-1],alist[j]
    print(alist)

# 将列表按照由大到小的顺序排列
def insert_sort_2(alist):
    n=len(alist)
    for i in range(n-1):
        for j in range(i+1,0,-1):
            if alist[j]>alist[j-1]:
                alist[j],alist[j-1]=alist[j-1],alist[j]
    print(alist)

alist=[5,3,1,2,-1,-9]
insert_sort_1(alist)
insert_sort_2(alist)
```

```python
# 将列表按照由小到大的顺序排列
def shell_sort_1(alist):
    n=len(alist)
    for id, gap in enumerate([2,1]):
        for i in range( len(range(id, n, gap))-1 ):
            for j in range(i*gap+id+gap,id,-gap):
                if alist[j]<alist[j-gap]:
                    alist[j],alist[j-gap]=alist[j-gap],alist[j]

    print(alist)

# 将列表按照由大到小的顺序排列？

alist=[5,3,1,2,-3,-7,-1000,5,6,7,8]
# alist=[-1,-4]
shell_sort_1(alist)
```
